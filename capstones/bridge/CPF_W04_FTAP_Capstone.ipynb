```python
import numpy as np
import pandas as pd

np.set_printoptions(precision=6, suppress=True)

pd.set_option(
    "display.float_format",
    lambda x: f"{x:.6f}"
)
```

## Market Specification and Assumptions

I construct a one-period market with three future states,

$$
\Omega=\{\omega_1,\omega_2,\omega_3\},
$$

and three traded assets: one risk-free asset and two risky assets.

The risk-free asset costs $1$ today and pays $1$ in every future state. This corresponds to a one-period risk-free gross return of

$$
R=1.
$$

The two risky assets have different state-contingent payoff patterns so that their payoffs, together with the risk-free asset, can potentially span the three-state payoff space.

I use the following market:

| Asset | Current Price | $\omega_1$ Payoff | $\omega_2$ Payoff | $\omega_3$ Payoff |
|---|---:|---:|---:|---:|
| Risk-free asset | $1.00$ | $1.00$ | $1.00$ | $1.00$ |
| Risky Asset 1 | $0.70$ | $2.00$ | $1.00$ | $0.00$ |
| Risky Asset 2 | $1.80$ | $0.00$ | $1.00$ | $3.00$ |

I will first solve for the state-price vector and verify that it is strictly positive. I will then check the rank of the payoff matrix to determine whether the market is complete.

```python
asset_names = [
    "Risk-free",
    "Risky Asset 1",
    "Risky Asset 2"
]

S0 = np.array([
    1.00,
    0.70,
    1.80
])

S1 = np.array([
    [1.00, 2.00, 0.00],
    [1.00, 1.00, 1.00],
    [1.00, 0.00, 3.00]
])

market = pd.DataFrame(
    S1,
    index=["omega_1", "omega_2", "omega_3"],
    columns=asset_names
)

market.loc["Current price"] = S0

display(market)
```

## State Prices, No-Arbitrage, and Completeness

I determine the state-price vector by requiring the current prices of all three traded assets to equal the state-price-weighted values of their future payoffs.

If

$$
\pi
=
\begin{pmatrix}
\pi_1 \\
\pi_2 \\
\pi_3
\end{pmatrix},
$$

then the state prices must satisfy

$$
S_0
=
S_1^\top \pi.
$$

For this market, the pricing equations are

$$
\pi_1+\pi_2+\pi_3=1,
$$

$$
2\pi_1+\pi_2=0.70,
$$

and

$$
\pi_2+3\pi_3=1.80.
$$

A strictly positive solution for all three state prices is evidence of no-arbitrage in this finite-state market.

I also compute the rank of the payoff matrix. Because there are three states, the market is complete if

$$
\operatorname{rank}(S_1)=3.
$$

If the payoff matrix has full rank, the three traded assets span the entire three-dimensional payoff space, so every contingent claim can be replicated uniquely.

```python
state_prices = np.linalg.solve(
    S1.T,
    S0
)

rank_S1 = np.linalg.matrix_rank(S1)

print("State prices:")
for i, value in enumerate(state_prices, start=1):
    print(f"pi_{i} = {value:.6f}")

print("\nAll state prices strictly positive:",
      np.all(state_prices > 0))

print("Payoff matrix rank:", rank_S1)
print("Number of states:", S1.shape[0])
print("Market complete:", rank_S1 == S1.shape[0])
```

### Interpretation

The state-price vector is

$$
\pi
=
\begin{pmatrix}
0.20 \\
0.30 \\
0.50
\end{pmatrix}.
$$

All three state prices are strictly positive. In this finite-state setting, the existence of a strictly positive state-price vector is consistent with the absence of arbitrage.

Because the risk-free gross return is $R=1$, the state prices also sum to one,

$$
0.20+0.30+0.50=1,
$$

so they coincide numerically with the corresponding risk-neutral probabilities in this example.

The payoff matrix has rank

$$
\operatorname{rank}(S_1)=3,
$$

which equals the number of states. The traded asset payoffs therefore span the entire three-state payoff space.

I classify the market as complete. Under the second Fundamental Theorem of Asset Pricing, this implies that the risk-neutral measure is unique. As a result, every contingent claim in this market can be replicated and has a unique no-arbitrage price.

## Contingent Claim 1

I first consider a claim that pays

$$
H^{(1)}
=
\begin{pmatrix}
1 \\
0 \\
0
\end{pmatrix}.
$$

The claim pays one unit only in state $\omega_1$.

Using the state-price vector,

$$
\pi
=
\begin{pmatrix}
0.20 \\
0.30 \\
0.50
\end{pmatrix},
$$

the no-arbitrage price is

$$
P\left(H^{(1)}\right)
=
\pi^\top H^{(1)}
=
0.20.
$$

Because the market is complete, I also expect this claim to be exactly replicable by the three traded assets. I solve

$$
S_1\theta^{(1)}=H^{(1)}
$$

and compare the initial cost of the replicating portfolio with the state-price valuation.

```python
H1 = np.array([
    1.0,
    0.0,
    0.0
])

price_H1_state = state_prices @ H1

theta_H1 = np.linalg.solve(
    S1,
    H1
)

replicated_H1 = S1 @ theta_H1

price_H1_rep = S0 @ theta_H1

print(f"State-price value: {price_H1_state:.6f}")

print("\nReplicating portfolio weights:")
for name, weight in zip(asset_names, theta_H1):
    print(f"{name}: {weight:.6f}")

print("\nReplicated payoff:", replicated_H1)

print(
    "Replication cost:",
    f"{price_H1_rep:.6f}"
)
```

### Interpretation

For the first contingent claim,

$$
H^{(1)}
=
\begin{pmatrix}
1 \\
0 \\
0
\end{pmatrix},
$$

the state-price valuation is

$$
P\left(H^{(1)}\right)=0.20.
$$

The replicating portfolio is

$$
\theta^{(1)}
=
\begin{pmatrix}
-3 \\
2 \\
1
\end{pmatrix},
$$

corresponding to a short position of three units in the risk-free asset, two units in Risky Asset 1, and one unit in Risky Asset 2.

Its future payoff is

$$
S_1\theta^{(1)}
=
\begin{pmatrix}
1 \\
0 \\
0
\end{pmatrix},
$$

which exactly matches the claim.

The initial cost of the replicating portfolio is

$$
S_0^\top\theta^{(1)}
=
0.20,
$$

which is identical to the state-price valuation.

This equality confirms the complete-market pricing logic: the claim is attainable, and both state-price valuation and replication produce the same unique no-arbitrage price.

## Contingent Claim 2

I next consider a second contingent claim with payoff

$$
H^{(2)}
=
\begin{pmatrix}
0 \\
1 \\
2
\end{pmatrix}.
$$

Using the state-price vector,

$$
\pi
=
\begin{pmatrix}
0.20 \\
0.30 \\
0.50
\end{pmatrix},
$$

the state-price valuation is

$$
P\left(H^{(2)}\right)
=
\pi^\top H^{(2)}
=
0.20(0)+0.30(1)+0.50(2).
$$

Because the market is complete, I also solve

$$
S_1\theta^{(2)}=H^{(2)}
$$

to construct the unique replicating portfolio and compare its initial cost with the state-price valuation.

```python
H2 = np.array([
    0.0,
    1.0,
    2.0
])

price_H2_state = state_prices @ H2

theta_H2 = np.linalg.solve(
    S1,
    H2
)

replicated_H2 = S1 @ theta_H2

price_H2_rep = S0 @ theta_H2

print(f"State-price value: {price_H2_state:.6f}")

print("\nReplicating portfolio weights:")
for name, weight in zip(asset_names, theta_H2):
    print(f"{name}: {weight:.6f}")

print("\nReplicated payoff:", replicated_H2)

print(
    "Replication cost:",
    f"{price_H2_rep:.6f}"
)
```

### Interpretation

For the second contingent claim,

$$
H^{(2)}
=
\begin{pmatrix}
0 \\
1 \\
2
\end{pmatrix},
$$

the state-price valuation is

$$
P\left(H^{(2)}\right)=1.30.
$$

The replicating portfolio is

$$
\theta^{(2)}
=
\begin{pmatrix}
2 \\
-1 \\
0
\end{pmatrix}.
$$

This corresponds to holding two units of the risk-free asset, shorting one unit of Risky Asset 1, and taking no position in Risky Asset 2.

The portfolio payoff is

$$
S_1\theta^{(2)}
=
\begin{pmatrix}
0 \\
1 \\
2
\end{pmatrix},
$$

which exactly matches the claim.

The initial cost of the replicating portfolio is

$$
S_0^\top\theta^{(2)}
=
1.30,
$$

which again matches the state-price valuation.

As with the first claim, the agreement between state-price valuation and replication reflects the completeness of the market and the uniqueness of the no-arbitrage price.

## Conclusion and FTAP Interpretation

The constructed three-state market admits the strictly positive state-price vector

$$
\pi
=
\begin{pmatrix}
0.20 \\
0.30 \\
0.50
\end{pmatrix}.
$$

The existence of strictly positive state prices is consistent with the absence of arbitrage. In this one-period finite-state setting, this illustrates the first Fundamental Theorem of Asset Pricing: no-arbitrage is associated with the existence of a strictly positive pricing measure.

The payoff matrix has full rank,

$$
\operatorname{rank}(S_1)=3,
$$

which equals the number of states. The traded assets therefore span the entire payoff space, so the market is complete.

This illustrates the second Fundamental Theorem of Asset Pricing: under no-arbitrage, completeness is associated with uniqueness of the risk-neutral measure.

Both contingent claims were exactly replicated by traded-asset portfolios, and in each case the replication cost matched the state-price valuation:

$$
P\left(H^{(1)}\right)=0.20,
$$

and

$$
P\left(H^{(2)}\right)=1.30.
$$

The capstone therefore demonstrates the complete-market FTAP logic directly: positive state prices support no-arbitrage, full spanning gives completeness, and completeness yields unique no-arbitrage prices for contingent claims.

```python
# 1. Market-dimension validation
assert S1.shape == (3, 3)
assert S0.shape == (3,)

print("PASS 1 — Market dimensions are correct.")


# 2. State-price validation
assert np.all(np.isfinite(state_prices))
assert np.all(state_prices > 0)
assert np.allclose(S1.T @ state_prices, S0)

print("PASS 2 — State prices are strictly positive and price all traded assets correctly.")


# 3. Completeness validation
assert np.linalg.matrix_rank(S1) == 3

print("PASS 3 — Payoff matrix has full rank; the market is complete.")


# 4. Claim 1 replication and pricing validation
assert np.allclose(S1 @ theta_H1, H1)
assert np.isclose(price_H1_state, state_prices @ H1)
assert np.isclose(price_H1_rep, S0 @ theta_H1)
assert np.isclose(price_H1_state, price_H1_rep)

print("PASS 4 — Claim 1 is replicated exactly and both pricing methods agree.")


# 5. Claim 2 replication and pricing validation
assert np.allclose(S1 @ theta_H2, H2)
assert np.isclose(price_H2_state, state_prices @ H2)
assert np.isclose(price_H2_rep, S0 @ theta_H2)
assert np.isclose(price_H2_state, price_H2_rep)

print("PASS 5 — Claim 2 is replicated exactly and both pricing methods agree.")


# 6. Expected numerical results
assert np.allclose(
    state_prices,
    np.array([0.2, 0.3, 0.5])
)

assert np.isclose(price_H1_state, 0.2)
assert np.isclose(price_H2_state, 1.3)

print("PASS 6 — Key numerical results match the stated capstone results.")


print("\nALL FTAP CAPSTONE VALIDATION CHECKS PASSED.")
```
